```python
import numpy as np
import pandas as pd

print("✅ NumPy and Pandas imported successfully")
print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
category_translation = pd.read_csv("product_category_name_translation.csv")

print("All Olist datasets loaded successfully")
```

```python
# Check shape of each dataset

datasets = {
    "orders": orders,
    "order_items": order_items,
    "customers": customers,
    "products": products,
    "payments": payments,
    "reviews": reviews,
    "sellers": sellers,
    "geolocation": geolocation,
    "category_translation": category_translation
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
#Check column names of all tables
for name, df in datasets.items():
    print(f"\n===== {name.upper()} =====")
    print(df.columns.tolist())
```

```python
# Convert timestamp columns in orders table

date_columns_orders = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns_orders:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

print(" Orders date columns converted successfully")
```

```python
print("Minimum order purchase date:", orders["order_purchase_timestamp"].min())
print("Maximum order purchase date:", orders["order_purchase_timestamp"].max())
```

```python
# Null value check

for name, df in datasets.items():
    print(f"\n===== NULL VALUES IN {name.upper()} =====")
    print(df.isnull().sum())
```

```python
# Duplicate row check

for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()
    print(f"{name}: {duplicate_count} duplicate rows")
```

```python
summary = []

for name, df in datasets.items():
    summary.append({
        "table_name": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicate_rows": df.duplicated().sum(),
        "total_null_values": df.isnull().sum().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```
